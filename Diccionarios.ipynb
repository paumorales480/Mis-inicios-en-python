#### Diccionarios 

¿Qué es? 
Es una estructura de datos mutable y o ordenada que permite almacenar pares de clave-valor. Cada elemento en un diccionario consiste en una clave única y su valor correspondiente. Los diccionarios se encierran entre llaves {}, y los pares clave-valor se separan por comas.

```python
#======= EJEMPLO ===============
persona={"nombre": "Paulina", "edad": 24}
print("Nombre de la persona:",persona["nombre"])
print("Edad de la persona:",persona["edad"])
```

```python
#Otra forma de conocer los valores: 

print("nombre de la persona:", persona.get("nombre"))
```

```python
#==================== MÉTODOS DE DICCIONARIOS ==========================
""" Métodos para manipular y acceder a elementos de un diccionario
keys(): Devuelve una vista de todas las claves del diccionario. 
values(): Devuelve una vista de todos los valores del diccionario. 
items(): Devuelve una vista de todos los pares clave-valor del diccionario.
update(otro diccionario): Actualiza el diccionario con los pares clave-valor de otro diccionario"""

#  EJEMPLO: 

#Creamos un nuevo diccionario 

personas = {"Nombre":"Paulina", "edad":24, "ciudad":"cdmx", "carrera": "medicina"}
print(personas)
print(personas["Nombre"])
```

```python
#Uso de keys
print(personas.keys()) #Salen todas las claves de mi diccionario

#Uso de values: 

print(personas.values()) # Me da los valores en mi diccionario 

#Uso de items

print(personas.items()) # Vista de todos los pares clave-valor 

#Uso de update 

personas.update({"profesion":"economista"})
persona
print(personas)
```

```python
#Vamos a ver cómo eliminar claves de un diccionario 

# 1. Uso de "del"
del personas["carrera"]
print(personas.items())

#Con pop

personas.pop("edad")
print("diccionario modificado:", personas)

eliminado = personas.pop("profesion")
print(eliminado)
```
