```r
talleres_raw <- googlesheets4::read_sheet(sheet_key, sheet = 1,
                                      col_types = "c") %>%
  janitor::clean_names() %>%
  select(1:19)
```

```r
talleres_raw <- talleres_raw %>%
clean_names() %>%
  
  # Pasar a formato largo
  
  pivot_longer(
    cols = c(
      medio_ambiente_agua_y_riesgos_problematicas,
      medio_ambiente_agua_y_riesgos_posibles_soluciones,
      medio_ambiente_agua_y_riesgos_opiniones,
      espacio_publico_vivienda_movilidad_y_democracia_problematicas,
      espacio_publico_vivienda_movilidad_y_democracia_posibles_soluciones,
      espacio_publico_vivienda_movilidad_y_democracia_opiniones,
      seguridad_salud_mujeres_problematicas,
      seguridad_salud_mujeres_posibles_soluciones,
      seguridad_salud_mujeres_opiniones,
      cultura_educacion_ciencia_economia_y_trabajo_problematicas,
      cultura_educacion_ciencia_economia_y_trabajo_posibles_soluciones,
      cultura_educacion_ciencia_economia_y_trabajo_opiniones
    ),
    names_to = "variable",
    values_to = "texto"
  ) %>%
  
  filter(!is.na(texto), texto != "") %>%
  
  separate_rows(texto, sep = ";") %>%
  
  mutate(texto = str_trim(texto)) %>%
  
  mutate(
    tipo = case_when(
      str_detect(variable, "problematicas") ~ "problematica",
      str_detect(variable, "posibles_soluciones") ~ "solucion",
      str_detect(variable, "opiniones") ~ "opinion"
    ),
    tema = case_when(
      str_detect(variable, "medio_ambiente") ~ "Medio ambiente, agua y riesgos",
      str_detect(variable, "espacio_publico") ~ "Espacio público, vivienda, movilidad y democracia",
      str_detect(variable, "seguridad_salud") ~ "Seguridad, salud, mujeres",
      str_detect(variable, "cultura_educacion") ~ "Cultura, educación, ciencia, economía y trabajo"
    )
  ) %>%
  
  select(-variable)
```

```r

talleres_raw <- talleres_raw %>%
  mutate(
    texto = texto %>%
      stri_trans_general("Latin-ASCII") %>%   # quitar acentos
      str_to_lower() %>%                      # minúsculas
      str_replace_all("[[:punct:]]", "") %>%  # quitar puntuación
      str_squish() %>%                        # quitar espacios extra
      str_split("\\s+") %>%                   # separar palabras
      map(~ .x[!.x %in% stopwords]) %>%       # quitar stopwords
      map_chr(paste, collapse = " ")          # reconstruir texto
  ) %>%
  filter(!is.na(texto), texto != "")

head(talleres_raw, 5)
```
