#### Recuerdo de mis primeras listas

Una lista es mutable y ordenada y sus elementos pueden ser de diferentes tipos de datos

```python
# Vamos a crear una lista

enteros = [1,2,3,4,5]
print(enteros)

# Otra forma de hacerlo más cool:

entero_cool = list(range(1,51)) #No olvidar que el primero se incluye y el segundo nop
print(entero_cool)
```

```python
#¿Cómo imprimir elementos específicos?
print(entero_cool[-1]) #último elemento
print(entero_cool[0]) #primero
print(entero_cool[0:5]) #Primeros cinco
```

```python
#¿Qué pasa si quiero imprirmir los elementos que son multiplos de X número?

#Primera forma
#IMPRIMIR MÚLTIPLOS DE 5
for multiplo in entero_cool:
    if multiplo % 5 == 0: # % -> DEVUELVE EL RESTO
        print(multiplo)

#Segunda forma:

multiplos = [multiplo for multiplo in entero_cool if multiplo % 5 == 0 ]
print(multiplos)

#multiplos = [multiplo        for multiplo in entero_cool   if multiplo % 5 == 0]
#                ↑               ↑                              ↑
#        qué guardás      recorrés la lista            solo si cumple esto
```

```python
# Lista de tipo string

things = ["libro", "celular", "reloj", "pincel", "museos", "arco", "comida", "Libro"]
```

```python
#Imprimir el último
print(things[-1])
#Imprimir los primeros 3
print(things[0:3])
#Imprimir todos menos el último
print(things[:6])
#Imprimir todos menos el primero
print(things[1:])
```

```python
# qué tal si quiero los que inicien con X letra?
#PRIMERA FORMA:

for thing in things:
    if thing.startswith("c"):
        print(thing)

```

```python
#SEGUNDA FORMA:
cosas_conc = [thing for thing in things if thing.startswith("c")]
print(cosas_conc)
```

```python
#qué tal si quiero saber cuántas veces aparece cierta palabra?
#Para eso está el poderosísimo "count"

things.count("libro")

#Si quiero que cuente teniendo minúsculas o mayúsculas:
#1. Debo limpiar mi lista

things_clean = [thing.lower() for thing in things]
print(things_clean)

```

```python
things_clean.count("libro")
```

#### Lista de listas :)

```python
integrantes = [["Fer", 24, "Relaciones Internacionales"], ["Ceci", 27, "Derecho"],
["Luis", 32, "Relaciones internacionales"], ["Rosy", 31, "Relaciones internacionales"]]

print(integrantes)
```

```python
#Edad de Ceci
print(integrantes[1][1])

#Primer persona, toda su info
print(integrantes[0])

#Solo el nombre de todos mis integrantes
print("INTEGRANTES:")
for integrante in integrantes:
    print(integrante[0])

#Solo las edades de los integrantes
print("edades:")

for edades in integrantes:
    print(edades[1])
```

```python
#Si yo quisiera convertir o solo jalar las edades a una nueva lista:
edades_integrantes = [edades[1] for edades in integrantes]
print(edades_integrantes)
```

```python
import statistics

print("el promedio de la edad es:", statistics.mean(edades_integrantes))
```

```python
#Diferencia ente len y count

print(len(things))        # 5 → cuántos elementos hay en total
print(things.count("libro"))  # 3 → cuántas veces aparece "libro"
```

#### Métodos para manipular y modificar listas

```python
# Append -> agrega un elemento al final de las lista
#Insert -> inserta un elemento en una posición específica de la lista
#Remove -> elimina la primera aparición de un elemento en la lista
#pop -> elmina y devuelve el elemento en una posición específica de la lista
#sot-> ordena los elementos de la lista en orden ascendente
#reverse -> invierte el orden de los elementos de la lista

##### EJEMPLOS #####

colores = ["amarillo", "rojo", "azul", "naranja", "rosa", "dorado"]

#Agrego "negro"
colores.append("negro")

print("último elemento:", colores[-1])

# Insertaré "morado" en el segundo elemento

colores.insert(1,"morado") #coloca en el index 1 el elemento "morado"
print("segundo elemento:", colores[1])

#Voy a eliminar "azul"
colores.remove("azul")

#puedo rectificar si está o no en la lista
if "azul" in colores:
    print("azul está en la lista")
else:
    print("azul no está en la lista")

# Uso de pop
print(colores)
print("uso de pop:")
colores.pop(1)
print(colores)

#veo si "rojo" sigue en mi lista

if "rojo" in colores:
  print("rojo está en la lista")
else:
    print("rojo no está en la lista")

#uso de sort (acomoda en orden alfabético)
print(colores)
colores.sort()
print("reacomodo:", colores)

#Uso de reverse
colores.reverse()
print("reversa:", colores)
```

```python
#========= Listas de comprensión ===========
#Son una forma concisa de crear nuevas listas basadas en
#una secuencia existente. Permite filtrar y transformar los elementos
#de una lista en una sola línea de código.

#EJEMPLO:
numeros =[10,30,100,2]
cuadrados = [x **2 for x in numeros if x%2 ==0]
#Traducción jiij:
#lista "cuadrados", la variable x se eleva al cuadro para todas las variables en
#"números" que son divisibles entre 2
print(cuadrados)
```
