```python
#Funciones con número variable de argumentos 

def media_numeros(*numeros):
        total = sum(numeros)
        cantidad = len(numeros)
        media=total/cantidad
        return media


print("Media1:", media_numeros(2,3,4,3,3,4,2,1,2,3))
print("Media2:", media_numeros(10,10,9,2,10))
```

```python
"""Crea una función llamada calculadora que reciba dos números y una operación ("suma", "resta", "multiplicacion", "division") e imprima el resultado."""

def calculadora (x, y, operacion):
    if operacion == "suma": 
        return (x+y)
    elif operacion == "resta":
        return(x-y)
    elif operacion == "producto":
        return(x*y)
    elif operacion == "division":
        return(x/y)
```

```python
print("5X5=", calculadora(5,5,"producto"))
print("25+25=", calculadora(25, 25,"suma"))
```

#### Ejercicio 2: Analizador de lista
Crea una función llamada analizar que reciba una lista de números e imprima:

El total de elementos
La suma
El promedio
El número mayor
El número menor



#### Si se Usa "*" son números sueltos y si no lo tiene, acepta una lista

```python
def analizar (*numeros, analisis):
    if analisis == "total":
          return len(numeros)
    elif analisis == "suma":
          return sum(numeros)
    elif analisis== "promedio":
          return sum(numeros)/len(numeros)
    elif analisis == "mayor":
          return max(numeros)
    elif analisis == "menor":
          return min(numeros)
    
print("resultado", analizar(2,3,4,3,2,3, analisis = "suma"))

print("mínimo:", analizar(10,100,2000,200030, analisis= "mayor"))
```

```python
# Que me imprima todo: total, suma, promedio, mayor y menor
def analisis (*numeros):
    total = len(numeros)
    suma = sum(numeros)
    promedio = suma/total
    mayor = max(numeros)
    menor=min(numeros)
    return total, suma, promedio, mayor, menor

print("Resumen", analisis(2,3,4,4,2,1))
```
