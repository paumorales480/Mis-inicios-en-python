#### ¿Qué onda con las tuplas?

Una tupla es una estructura de datos inmutable y ordenada que permite almacenar una colección de elementos. Los elementos de una tupla se encierran entre paréntesis (), separados por comas.

```python
mi_primer_tupla = (1,2,3,4,5)

#Para visualizar elementos, es parecido a las listas

#Primer elemento: 
print("primer elemento:",mi_primer_tupla[0])

#"Último elemento"
print("último elemento:", mi_primer_tupla[-1])
```

```python
#métodos para trabajar con ellas:

#count -> Número de veces que aparece un elemento en la tupla
#index(elemento) -> devuelve el índice de la primera aparición de un elemento en la tupla
#len(tupla) -> devuelve la longitud de la tupla

#EJEMPLO

segunda = (2,2,2,3,4,5,3,1)
print(segunda.count(2))
print(segunda.index(4))
print(segunda.index(2,0,2)) #devuelve el índice, no el conteo
#                  ↑  ↑  ↑
#              qué  desde  hasta
#            buscás  dónde  dónde

#¿cuántos dos hay entre la posición 0 y 4?
segunda[0:5].count(2)
```
