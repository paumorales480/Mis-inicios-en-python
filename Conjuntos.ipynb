#### Conjuntos

Es una estructura de datos mutable y no ordenada que permite almacenar una colección de elementos únicos. Se encierran entre {} o con la función set

```python
# Creación de conjuntos

nombres = {"Paulina", "Fernanda", "Adriana"}
numeros = set([3,4,5,2,3,4])
```

```python
# Con los conjuntos de pueden hacer operaciones matemáticas:

# Unión : |
# Intersección: &
#Diferencia: - 
#Diferencia simétrica: ^

#    EJEMPLOS

a={1,2,3,4,5}
b={2,3,6,9,8}

union= a|b
print("Unión de conjunto a y b:", union)

interseccion=a&b
print("Intersección a y b:", interseccion)

diferencia=a-b
print("diferencia entre conjunto a y b:", diferencia)

#Nota porque qué onda que no me acordaba:

""" La diferencia simétrica es el conjunto de elementos que pertenecen a A o a B, pero 
no a ambos simultáneamente"""

dif_sime=a^b
print("Diferencia simétrica:", dif_sime)
```

```python
# Métodos de conjuntos. Para manipular y acceder a los elementos.

# add(): agrega un elemento al conjunto
# remove(): elminina un elemento del conjunto. Si el elemento no existe, genera un error
# discard(): elimina un elemento del conjunto si está presente. Si el elemento no existe, no hace nada.
# clear(): elimina todos los elementos del conjunto

# EJEMPLO:

frutas = {"manzana", "uva", "naranja"}

#Add
frutas.add("pera")
print(frutas)

#Remove
frutas.remove("pera")
print(frutas)

#discard -> la diferencia con remove es que con discard, no truena 
frutas.discard("pera")
print(frutas)

#clear
frutas.clear()
print(frutas)
```
