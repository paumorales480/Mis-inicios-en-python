```python
# Algunos operadores aritméticos que suelo olvidar en python: 

# // -> Divide y devuelve un resultado de tipo entero (se descarta la parte decimal)
# % -> Devuelve el resto de la división entre el primer valor y el segundo
# ** -> Exponenciación

## EJEMPLOS: 

a = 10 
b = 3

print("División:", a//b)
print("Módulo:", a%b)
print("Exponenciación:", a**b)
```

```python
# Comparación
m = 10
f=20
print("¿m<=f?", m<=f)
print( "¿f<m?", f<m)
print("¿f>m?", f>m)
```

```python
# ===== ESTRUCTURAS DE CONTROL =====

# if -> Se utiliza para ejecutar un bloque de código si una condición es verdadera

#If-else -> nos permite especificar un bloque de código alternativo que se ejecutará 
# si la condición de  if es falsa.

# elif -> Es una combinación de else+if, o sea "Si lo anterior no se cumplió, prueba esta nueva condición"


# If-elif-else -> 
#Ejemplo: 
edad = 14
if edad >= 18: 
    print("Eres mayor de edad")
else:
    print("Eres menor de edad") 
```

```python
#Otro ejemplo usando las tres

edad = 30

if edad <= 17: 
    print("Menor de edad")
elif edad <= 29: 
    print("Joven")
elif edad <= 39:
    print("menor de 40 años")
else: 
    print("Tiene 40 o más")
```

```python
###### Bucles/loops ######

# FOR: Lo que se está haciendo es iterar. Es decir, repetir una acción elemento
#por elemento sobre una colección de datos 

frutas = ["manzana", "uva", "piña"]

for fruta in frutas:
    print(fruta)

# WHILE: Se utiliza para repetir un bloque de código mientras una concidión sea verdadera


contador = 1 
while contador <= 5:
    print(contador)
    contador += 1
# Si la condición nunca se vuelve falsa, el bucle se ejecutará indefinidamente -> bucle infinito
```

```python
# Más ejemplos:
print("Números del 1 al 5 multiplicados por 2 con bucle for:")
for numero in range(1,6): #es 6 porque no se incluye for default el segundo pero sí el primero
    print(numero*2)

print("Se puede hacer el mismo ejecicio con while:")

numero_2 = 1
while numero_2 <= 5:
    print(numero_2*2)
    numero_2+=1
```

```python
#¿Qué onda con BREAK?

# Break se utiliza para salir prematuramente de un bucle, independientemente de la condición. 
#Cuando se encuentra un break, el bucle se detiene y el flujo de ejecución continúa con la siguiente instrucción fuera del bucle

contador = 0 
while True:
    print(contador) 
    contador += 1
    if contador >= 11: 
        break
```

```python
#¿Qué onda con CONTINUE?
# Continue -> Se utiliza para saltar el resto del bloque de código dentro de un bucle y para a la siguiente iteración 

for i in range(10):
    if i % 2 == 0:
        continue
    print(i)

#Mi mente lo lee como: para la variable i, en un rango del 0 al 9. Si i al dividirlo entre 2
#y da un residuo de 0 (divisible entre 2), continúa la iteración e imprimer i
```

```python
# ¿Qué onda con PASS? 

#PASS -> es una operación nula que no hace nada. Se utiliza como marcador de posición
#cuando de requiere una instrucción sintácticamente, pero no se desea realizar ninguna acción

#Esto puede ser útil cuando se está desarrollando un programa y se desea reservar un bloque 
# de código para implementarlo más adelante.

for i in range(5): #no olvidar que no contempla al 5, significa que va del 0 al 4
    pass
```
